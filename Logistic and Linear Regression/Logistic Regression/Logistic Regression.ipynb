### Define Input Features 

```python
# Store the names of the input features in a list
# These columns will be used to predict whether a patient has diabetes
x = [
    'Pregnancies',
    'Glucose',
    'BloodPressure',
    'SkinThickness',
    'Insulin',
    'BMI',
    'DiabetesPedigreeFunction',
    'Age'
]
```

```python
# Store the name of the target/output variable
y = ['Outcome']
```

```python
# Create a new DataFrame df2 using the data from df1
df2 = pd.DataFrame(data=df1)

# Display the first five rows of the new DataFrame
df2.head()
```

### Split Dataset into Training and Testing Data

```python
# Import the train_test_split function from Scikit-learn
from sklearn.model_selection import train_test_split

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    # Use all columns except 'Outcome' as input features
    df1.drop('Outcome', axis=1),

    # Use 'Outcome' as the target variable
    df1['Outcome'],

    # Use 20% of the dataset for testing
    test_size=0.20,

    # Set a fixed random state so that the same split
    # is obtained every time the code is executed
    random_state=101
)
```

```python
# Display the first five rows of the testing input data
X_test.head()
```

### Logistic Regression Model

```python
# Import the LogisticRegression algorithm from Scikit-learn
from sklearn.linear_model import LogisticRegression

# Create a Logistic Regression model
# solver='lbfgs' specifies the optimization algorithm
# max_iter=7600 allows the model to perform up to 7600 iterations
LRModel = LogisticRegression(
    solver='lbfgs',
    max_iter=7600
)

# Train the Logistic Regression model using the training data
LRModel.fit(X_train, y_train)
```

```python
# Use the trained Logistic Regression model
# to predict diabetes outcomes for the test dataset
predictions_diabetes = LRModel.predict(X_test)
```

```python
# Import metrics used to evaluate the classification model
from sklearn.metrics import classification_report, confusion_matrix

# Generate and display the classification report
# It provides precision, recall, F1-score and support
print(
    classification_report(
        y_test,
        predictions_diabetes
    )
)
```
