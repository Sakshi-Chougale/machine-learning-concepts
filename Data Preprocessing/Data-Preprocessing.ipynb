```python
# Outlier detection in the 'Age' column using the IQR method

# Calculate the first quartile (Q1), which represents the 25th percentile
q1 = df['Age'].quantile(0.25)

# Display the calculated Q1 value
q1
```

```python
# Calculate the third quartile (Q3), which represents the 75th percentile
q3 = df['Age'].quantile(0.75)

# Display the calculated Q3 value
q3
```

```python
# Calculate the Interquartile Range (IQR)
# IQR is the difference between the third quartile (Q3) and first quartile (Q1)
iqr = q3 - q1

# Display the calculated IQR value
iqr
```

```python
# Calculate the lower boundary for detecting outliers
# Any value below this boundary can be considered a potential outlier
lower = q1 - 1.5 * iqr

# Display the calculated lower boundary
lower
```

```python
# Calculate the upper boundary for detecting outliers
# Any value above this boundary can be considered a potential outlier
upper = q3 + 1.5 * iqr

# Display the calculated upper boundary
upper
```

```python
# Select rows where Age is below the lower bound
# AND above the upper bound
df[(df['Age'] < lower) & (df['Age'] > upper)]
```

```python
#Outlier detection in 'Salary' column=IQR
q1=df['Salary'].quantile(0.25)
q1
```

```python
q3=df['Salary'].quantile(0.75)
q3
```

```python
iqr=q3-q1
iqr
```

```python
lower=q1-1.5*iqr
lower
```

```python
upper=q3+1.5*iqr
upper
```

```python
#outlier rows
df[(df['Salary']<lower) & (df['Salary']>upper)]
#no outliers detected using IQR.
```

# OneHot Encoding

```python
#Encoding Categorical datas using OneHotEncoding
df['Company'].unique()
```

```python
df['Place'].unique()
```

```python
df['Gender'].unique()
```

```python
# Use 'OneHotEncoder' to create dummy variables for each of the nominal categorical columns.
```

```python
# OneHotEncoding

from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder()

# Encode Company, Place and Gender
df_array = ohe.fit_transform(
    df[['Company', 'Place', 'Gender']]
).toarray()

# Get category names automatically
categories = ohe.categories_

# Convert categories into one list
single_list = [
    item
    for category in categories
    for item in category
]

# Create encoded DataFrame
df_new = pd.DataFrame(
    df_array,
    columns=single_list
)

# Combine original dataframe with encoded dataframe
df_ml = pd.concat([df, df_new], axis=1)

# Remove original categorical columns
df_ml.drop(
    ['Company', 'Place', 'Gender'],
    axis=1,
    inplace=True
)

# Final encoded dataset
df_ml
```
