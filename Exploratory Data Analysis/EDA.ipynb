## Basic Information 
1. Count of survivors and non-survivors

```python
titanic['Survived'].value_counts()

# This counts the number of passengers for each value in the survived column:
# 0 → did not survive
# 1 → survived
```

```python
sns.countplot(data=titanic, x='Survived')
# This creates a count plot showing the number of passengers in each survival category.
```

2.counts how many passengers belonged to each passenger class

```python
titanic['Pclass'].value_counts()
```

```python
p_class = titanic[['Pclass']]
p_class.head()
```

3.Counts how many genders are there in the dataset

```python
titanic['Sex'].value_counts()
```

```python
sns.countplot(data=titanic, x='Sex')
# This creates a count plot showing the number of male and female passengers
```

4.Show the passengers who were less than 20 years

```python
less_than_20 = titanic[titanic['Age'] < 20 ]

less_than_20.head()
```

```python
len(less_than_20)
#This counts how many passengers are under 20.
```

5. How many unique cities are there in embark_town?

```python
titanic['Embarked'].nunique()
# nunique() counts the number of unique values in the column.
```

```python
titanic['Embarked'].value_counts().plot(kind='bar')

# value_counts() counts how many passengers belong to each embarkation category.
# plot(kind='bar') displays those counts as a bar chart.
```

6. Missing Data

```python
titanic.isnull().sum()
# This checks every column and counts how many values are missing (NaN)
```

7.More Analysis

```python
sns.countplot(data=titanic, x='Survived', palette='autumn', hue='Sex')

# This creates a count plot showing survival status separated by gender.
# Survived → survival category
# Sex → male/female groups
# palette='autumn' → keeps the notebook's original color palette
# hue='Sex' → separates the bars according to gender
```

```python
sns.countplot(data=titanic, x='Survived', palette='viridis', hue='Pclass')
```

```python
titanic.Age.plot(kind='hist',bins=30,color='green')
# This plots the distribution of passenger ages using 30 bins.
```

```python
sns.scatterplot(data=titanic, x='Age', y='Fare', hue='Pclass')
# This creates a scatter plot showing the relationship between age and fare, separated by passenger class.
```

```python
sns.scatterplot(data=titanic, x='Age', y='Fare', hue='Sex')
# This shows the relationship between age and fare, separated by gender.
```

8.Checking corelating features

```python
correlation = titanic.corr(numeric_only = True)
correlation

# calculates the correlation between the numerical columns of the Titanic dataset.
```

```python
correlation['Survived']
# This selects the Survived column from the correlation table.
```

```python
plt.figure(figsize=(12,7))

sns.heatmap(correlation,annot=True,cmap='crest')    #displays the correlation values as a heatmap.
```
